```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("california_housing_train.csv")
plt.bar(df.index[:10], df["median_income"][:10])
plt.title("Median Income")
plt.show()
plt.bar(df.index[:10], df["median_house_value"][:10])
plt.title("Median House Value")
plt.show()
plt.barh(df.index[:10], df["total_rooms"][:10])
plt.title("Total Rooms")
plt.show()
```
